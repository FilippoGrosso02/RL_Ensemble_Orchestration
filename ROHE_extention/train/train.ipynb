```python
import os
import sys
from stable_baselines3 import PPO
import torch

current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)

sys.path.append(parent_dir)
from RL.eemls_env import EEMLSEnv
from simulation.profile_manager import ProfileManager



        
sim_config = {
    "config_path": "config/sim_config.yaml",
    "profile_path": "data/model_profile/model_profile.yaml",
    "label_path": "data/raw_sample/file_label.csv",
    "model_record_path": "data/record",
    "output_path": "results/eemls_inferences/"
}
env_config ={
    "sim_config": sim_config,
}

experiment_prefix = "ppo_run_"
run_id = 0
while True:
    result_dir = os.path.join(parent_dir, f"results/{experiment_prefix}{run_id}")
    if not os.path.exists(result_dir):
        experiment_id = f"{experiment_prefix}{run_id}"
        break
    run_id += 1
    
    
weights = {
    "accuracy": 1.0,
    "confidence": 0.0,
    "latency": 0.0,
    "energy": 0.0,
    "explainability": 0.0
}

env_config ={
    "sim_config": sim_config,
    "experiment_id": experiment_id,
    "num_models": 3,
    "max_models": 3,
    "min_models": 2,
    "num_inferences": 25,
    "weights": weights,
    "contract_path": "config/contract.yaml",
}

rl_env = EEMLSEnv(env_config)
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

MODEL_PATH = f"simulation_policy_{run_id}.zip"
log_dir = "logs"
if os.path.exists(MODEL_PATH):
    print("Loading existing model...")
    # model = PPO.load(MODEL_PATH, env=env, device=device)
    model = PPO.load(MODEL_PATH, env=rl_env)
else:
    print("No saved model found. Training a new one...")
    model = PPO("MlpPolicy", rl_env, verbose=1, n_steps=256, device=device, tensorboard_log=log_dir, learning_rate=0.0005)

# Training agent

if not os.path.exists(f"saved_model_{run_id}"):
    os.makedirs(f"saved_model_{run_id}")
for i in range(20):
    model.learn(total_timesteps=1024, reset_num_timesteps=False, tb_log_name=f"PPO_{run_id}")
    # create folder for saving model with run_id
    
    # Save the trained model
    model.save(f"saved_model_{run_id}/model_{run_id}_{i}")
```
