```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from managers.enviroment_manager import SimulationEnv
import os
import torch
import time
import shutil

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

MODEL_PATH = "simulation_policy.zip"

config_dir =  "profile/config_profile"
def clean_dir(config_dir):
    if os.path.exists(config_dir):
        try:
            shutil.rmtree(config_dir)  
            print("directory clean")
        except Exception as e:
            print(f"Error removing {config_dir}: {e}")
    else:
        print("doesnt exist")
        
    os.makedirs(config_dir, exist_ok=True)
clean_dir(config_dir)

# Create a vectorized environment with 4 parallel instances
def create_env():
    return SimulationEnv("ppo_test1", num_inferences=100, parallel_workers=1)

env = make_vec_env(create_env, n_envs=4)

# Load an existing model or create a new one
if os.path.exists(MODEL_PATH):
    print("Loading existing model...")
    model = PPO.load(MODEL_PATH, env=env)
else:
    print("No saved model found. Training a new one...")
    model = PPO("MlpPolicy", env, verbose=1, n_steps=128)

start_time = time.time()

# Training agent
model.learn(total_timesteps=5120)

end_time = time.time()
elapsed_time = end_time - start_time
print(f"Training time: {elapsed_time:.2f} seconds")

# Save the trained model
# model.save("simulation_policy.zip")
```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from managers.enviroment_manager import SimulationEnv
import os
import time
import shutil

config = {
    "model_path": "simulation_policy.zip",
    "config_dir": "profile/config_profile",
    "n_envs": 4,
    "total_timesteps": 5120,
    "ppo_params": {
        "policy": "MlpPolicy",
        "n_steps": 128,
        "batch_size": 64,
        "learning_rate": 3e-4,
        "gamma": 0.99,
        "gae_lambda": 0.95,
        "clip_range": 0.2,
        "ent_coef": 0.01,
        "vf_coef": 0.5,
        "max_grad_norm": 0.5,
        "verbose": 1
    }
}

# Function to clean and recreate directory
def clean_dir(config_dir):
    if os.path.exists(config_dir):
        try:
            shutil.rmtree(config_dir)  
            print("Directory cleaned")
        except Exception as e:
            print(f"Error removing {config_dir}: {e}")
    else:
        print("Directory does not exist")
    os.makedirs(config_dir, exist_ok=True)
clean_dir(config["config_dir"])

def create_env():
    return SimulationEnv("ppo_test1", num_inferences=100, parallel_workers=1)

env = make_vec_env(create_env, n_envs=config["n_envs"])

policy = config["ppo_params"].pop("policy")
model = PPO(
    policy , env, **config["ppo_params"]
)

start_time = time.time()

model.learn(total_timesteps=config["total_timesteps"])

end_time = time.time()
elapsed_time = end_time - start_time
print(f"Training time: {elapsed_time:.2f} seconds")

```
